```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python

DATA_PATH = r'D:\Academic\ML\ModelTraining\Preprocessing\combined_v2.csv'
OUT_DIR = r'D:\Academic\ML\ModelTraining\EDA_Combined'
os.makedirs(OUT_DIR, exist_ok=True)

df = pd.read_csv(DATA_PATH)
```

```python

CLASSES = ['Normal Traffic', 'DoS', 'DDoS', 'Port Scan', 'Brute Force']

colors = {
    'Normal Traffic': '#4C8BB5',
    'DoS': '#E07B54',
    'DDoS': '#6BAE75',
    'Port Scan': '#C06BA8',
    'Brute Force': '#C9A84C'
}
```

```python

# --- 1. Figure 3.3: Correlation Heatmap ---
sample_df = df.sample(n=min(15000, len(df)), random_state=42)

num_cols = [
    c for c in df.select_dtypes(include=[np.number]).columns
    if c not in ('Attack Type', 'Source')
]

corr = sample_df[num_cols[:30]].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(
    corr,
    cmap='coolwarm',
    center=0,
    cbar_kws={'shrink': 0.8}
)
plt.title('Feature Correlation Matrix (sample n=15,000)', fontsize=14)
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/fig3_3_correlation_matrix.png', dpi=300)
plt.close()
```

```python

# --- 2. Figure 3.4: 4-Key Features Log1p Density Grid ---
key_feats = [
    'Packet Length Mean',
    'Flow Duration',
    'Flow Bytes/s',
    'Fwd Packet Length Mean'
]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for ax, feat in zip(axes.flatten(), key_feats):
    if feat in df.columns:
        for cls in CLASSES:
            subset = df[df['Attack Type'] == cls][feat].dropna()

            sns.kdeplot(
                np.log1p(subset.clip(lower=0)),
                ax=ax,
                label=cls,
                color=colors[cls],
                fill=True,
                alpha=0.2
            )

        ax.set_title(f'{feat} (Log1p scale)', fontsize=12)
        ax.set_xlabel('log1p(value)')
        ax.legend()

plt.tight_layout()
plt.savefig(f'{OUT_DIR}/fig3_4_key_feature_distributions.png', dpi=300)
plt.close()
```

```python

# --- 3. Figure 3.5: Packet Length Mean Boxplot ---
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x='Attack Type',
    y='Packet Length Mean',
    order=CLASSES,
    palette=colors,
    showfliers=False
)

plt.title('Packet Length Mean Distribution by Attack Class', fontsize=13)
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/fig3_5_packet_length_mean_boxplot.png', dpi=300)
plt.close()
```

```python

# --- 4. Figure 3.6: Flow Duration Boxplot ---
plt.figure(figsize=(9, 5))

df['log_duration'] = np.log1p(df['Flow Duration'].clip(lower=0))

sns.boxplot(
    data=df,
    x='Attack Type',
    y='log_duration',
    order=CLASSES,
    palette=colors,
    showfliers=False
)

plt.title('Flow Duration Distribution by Attack Class (Log Scale)', fontsize=13)
plt.ylabel('log1p(Flow Duration in µs)')
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/fig3_6_flow_duration_boxplot.png', dpi=300)
plt.close()
```

```python

# --- 5. Figure 3.1: Feature Distribution Across Attack Families ---
plt.figure(figsize=(10, 6))

for cls in CLASSES:
    subset = df[df['Attack Type'] == cls]['Packet Length Mean'].dropna()

    sns.kdeplot(
        np.log1p(subset.clip(lower=0)),
        label=cls,
        color=colors[cls],
        linewidth=2
    )

plt.title('Feature Distribution Across Attack Families (Log1p Scale)', fontsize=13)
plt.xlabel('log1p(Packet Length Mean)')
plt.legend()
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/fig3_1_feature_distribution.png', dpi=300)
plt.close()

print("All plots generated!")
```
