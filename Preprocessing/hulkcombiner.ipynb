```python
import pandas as pd

hulk = pd.read_csv('../../Tests/DoSHulk.csv')
hulk['Attack Type'] = 'DoS'
hulk['Source'] = 'Live-Hulk'
print(hulk.shape)
print(hulk['Attack Type'].value_counts())
```

```python
combined = pd.read_csv('combined.csv')
print("Before:", combined.shape, combined['Attack Type'].value_counts().to_dict())

combined = pd.concat([combined, hulk], ignore_index=True)
print("After:",  combined.shape, combined['Attack Type'].value_counts().to_dict())

combined.to_csv('combined_v2.csv', index=False)
```
