```python
import pandas as pd

input_file = "Preprocessing/cicids2017_preprocessed.csv"
output_map = {
    'DoS': 'DoS.csv',
    'DDoS': 'DDoS.csv',
    'Port Scan': 'PortScan.csv',
    'Brute Force': 'BruteForce.csv',
}

df = pd.read_csv(input_file)

for label, filename in output_map.items():
    subset = df[df['Attack Type'] == label]
    subset.to_csv(filename, index=False)
    print(f"{filename}: {len(subset)} rows written")
```
